```python
import pickle
import joblib
import numpy as np
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../../data")
MODELS_DIR = Path("../../models")

with open(DATA_DIR / "cfe_batch_lr_20.pkl", "rb") as f:
    cfe_lr = pickle.load(f)
with open(DATA_DIR / "cfe_batch_en_20.pkl", "rb") as f:
    cfe_en = pickle.load(f)

lr_pipeline = joblib.load(MODELS_DIR / "a-lr.pkl")
en_pipeline = joblib.load(MODELS_DIR / "a-en.pkl")

feature_cols = list(dataA.columns)
feature_cols = [c for c in feature_cols if "_id" not in c and c != "outcome"]


def check_flips(cfe_results, pipeline, X, feature_cols):
    flip_counts = []
    for patient_idx, cfe_list in enumerate(cfe_results):
        if cfe_list is None:
            flip_counts.append(None)
            continue
        original_pred = pipeline.predict(X.iloc[[patient_idx]][feature_cols])[0]
        cfe_arr = np.array(cfe_list)
        cfe_df = pd.DataFrame(cfe_arr, columns=feature_cols)
        cfe_preds = pipeline.predict(cfe_df)
        flipped = (cfe_preds != original_pred).sum()
        flip_counts.append(flipped)
    return flip_counts


X = dataA.drop(columns=[c for c in dataA.columns if "_id" in c], errors="ignore")

lr_flips = check_flips(cfe_lr, lr_pipeline, X, feature_cols)
en_flips = check_flips(cfe_en, en_pipeline, X, feature_cols)

lr_valid = [f for f in lr_flips if f is not None]
en_valid = [f for f in en_flips if f is not None]

print(f"LR: {len(lr_valid)} patients with CFEs")
print(f"  Mean flips per patient: {np.mean(lr_valid):.1f}/20")
print(f"  All 20 flipped: {sum(f == 20 for f in lr_valid)}")
print(f"  Some failed: {sum(f < 20 for f in lr_valid)}")
print(f"  None flipped: {sum(f == 0 for f in lr_valid)}")

print(f"\nEN: {len(en_valid)} patients with CFEs")
print(f"  Mean flips per patient: {np.mean(en_valid):.1f}/20")
print(f"  All 20 flipped: {sum(f == 20 for f in en_valid)}")
print(f"  Some failed: {sum(f < 20 for f in en_valid)}")
print(f"  None flipped: {sum(f == 0 for f in en_valid)}")
```

```python
def check_post_flip_confidence(cfe_results, pipeline, X, feature_cols):
    """For each patient, return mean confidence in flipped class across their CFEs."""
    confidences = []
    for patient_idx, cfe_list in enumerate(cfe_results):
        if cfe_list is None:
            confidences.append(None)
            continue
        original_pred = pipeline.predict(X.iloc[[patient_idx]][feature_cols])[0]
        target_class = 1 - original_pred  # the class CFEs are aiming for

        cfe_df = pd.DataFrame(np.array(cfe_list), columns=feature_cols)
        cfe_probs = pipeline.predict_proba(cfe_df)[:, target_class]
        confidences.append(cfe_probs)  # array of 20 probabilities per patient
    return confidences


lr_conf = check_post_flip_confidence(cfe_lr, lr_pipeline, X, feature_cols)
en_conf = check_post_flip_confidence(cfe_en, en_pipeline, X, feature_cols)

lr_means = np.array([np.mean(c) for c in lr_conf if c is not None])
en_means = np.array([np.mean(c) for c in en_conf if c is not None])

print("Per-patient mean post-flip confidence:")
print(f"\nLR — mean={lr_means.mean():.3f}, std={lr_means.std():.3f}")
print(f"  min={lr_means.min():.3f}, max={lr_means.max():.3f}")
print(f"  patients with mean conf <0.6: {(lr_means < 0.6).sum()}")
print(f"  patients with mean conf >0.9: {(lr_means > 0.9).sum()}")

print(f"\nEN — mean={en_means.mean():.3f}, std={en_means.std():.3f}")
print(f"  min={en_means.min():.3f}, max={en_means.max():.3f}")
print(f"  patients with mean conf <0.6: {(en_means < 0.6).sum()}")
print(f"  patients with mean conf >0.9: {(en_means > 0.9).sum()}")
```
